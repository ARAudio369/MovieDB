```python
df = spark.read \
    .option("multiLine", "true") \
    .json("abfss://bronze@moviedbstorage01.dfs.core.windows.net/movies/movies.json")

display(df)
```

```python
df.select("title", "genres.name").show(truncate=False)
```

```python
silver_movies = df.select(
    "id",
    "title",
    "original_title",
    "overview",
    "original_language",
    "release_date",
    "status",
    "runtime",
    "budget",
    "revenue",
    "popularity",
    "vote_average",
    "vote_count",
    "adult",
    "video",
    "homepage",
    "imdb_id",
    "poster_path",
    "backdrop_path",
    "tagline",
    "belongs_to_collection"
)

display(silver_movies)
```

```python
from pyspark.sql.functions import col, try_to_date

silver_movies = silver_movies.withColumn(
    "release_date",
    try_to_date(col("release_date"), "yyyy-MM-dd")
)
silver_movies.printSchema()
```

```python
from pyspark.sql.functions import col

silver_movies = silver_movies.withColumn(
    "collection_id",
    col("belongs_to_collection.id")
).withColumn(
    "collection_name",
    col("belongs_to_collection.name")
)
```

```python
silver_movies.printSchema()
```

```python
silver_movies = silver_movies.drop("belongs_to_collection")
silver_movies.printSchema()
```

```python
silver_arrays = df.select(
    "id",
    "title",
    "genres",
    "origin_country",
    "production_companies",
    "production_countries",
    "spoken_languages"
)

display(silver_arrays)
```

```python
from pyspark.sql.functions import explode, col

silver_movie_genres = df.select(
    col("id").alias("movie_id"),
    explode("genres").alias("genre")
).select(
    "movie_id",
    col("genre.id").alias("genre_id"),
    col("genre.name").alias("genre_name")
)

display(silver_movie_genres)
```

```python
from pyspark.sql.functions import explode, col

silver_movie_origin_countries = df.select(
    col("id").alias("movie_id"),
    explode("origin_country").alias("origin_country")
)

display(silver_movie_origin_countries)
```

```python
from pyspark.sql.functions import explode, col

silver_movie_production_companies = df.select(
    col("id").alias("movie_id"),
    explode("production_companies").alias("production_company")
).select(
    "movie_id",
    col("production_company.id").alias("production_company_id"),
    col("production_company.logo_path").alias("production_company_logo_path"),
    col("production_company.name").alias("production_company_name"),
    col("production_company.origin_country").alias("production_company_origin_country")
)

display(silver_movie_production_companies)
```

```python
from pyspark.sql.functions import explode, col

silver_movie_production_countries = df.select(
    col("id").alias("movie_id"),
    explode("production_countries").alias("production_country")
).select(
    "movie_id",
    col("production_country.iso_3166_1").alias("production_country_code"),
    col("production_country.name").alias("production_country_name"),
)

display(silver_movie_production_companies)
```

```python
from pyspark.sql.functions import explode, col

silver_movie_spoken_languages = df.select(
    col("id").alias("movie_id"),
    explode("spoken_languages").alias("spoken_language")
).select(
    "movie_id",
    col("spoken_language.english_name").alias("spoken_language_english_name"),
    col("spoken_language.iso_639_1").alias("spoken_lanuage_country_code"),
    col("spoken_language.name").alias("spoken_lanuage_name")
)

display(silver_movie_spoken_languages)
```

```python
#Check Null count for ID column
silver_movies.filter(
    col("id").isNull()
).count()
```

```python
#Check that distinct count is equal to non-distinct count
silver_movies.select("id").distinct().count() == silver_movies.count()
```

```python
#Check Null count on one of the exploded/unnested dataframes
silver_movie_genres.filter(
    col("movie_id").isNull()
).count()
```

```python
silver_movie_production_countries.write \
    .mode("overwrite") \
    .parquet(
        "abfss://silver@moviedbstorage01.dfs.core.windows.net/movie_production_countries/"
    )
```
